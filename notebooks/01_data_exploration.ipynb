```python
import pandas as pd
from pathlib import Path

# Locate the project folder
project_dir = Path.cwd().parent

# Locate the dataset
data_file = project_dir / "data" / "raw" / "adult.csv"

# Load the dataset
df = pd.read_csv(data_file)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Check for missing values represented by '?'

missing_symbols = (df == "?").sum()

print("Missing values represented by '?':")
print(missing_symbols[missing_symbols > 0])
```

```python
total_missing = (df == "?").sum().sum()

print("Total missing values:", total_missing)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
import numpy as np

# Replace '?' with proper missing values
df.replace("?", np.nan, inplace=True)

print(" '?' values replaced with NaN.")
```

```python
missing_values = df.isnull().sum()

print("Missing values after conversion:")
print(missing_values[missing_values > 0])
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

missing_report = missing_report[missing_report["Missing Count"] > 0]

print(missing_report)
```

```python
for column in ["workclass", "occupation", "native-country"]:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print("Missing values after imputation:")
print(df.isnull().sum().sum())
```

```python
# Check for duplicate records

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows

df = df.drop_duplicates()

print("Duplicates removed.")
print("New dataset shape:", df.shape)
```

```python
# Identify categorical columns

categorical_columns = df.select_dtypes(include="object").columns

print("Categorical columns:")
print(list(categorical_columns))
```

```python
# Display unique values for each categorical column

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].unique())
```

```python
# Remove leading and trailing spaces from categorical values

for column in categorical_columns:
    df[column] = df[column].str.strip()

print("Categorical values standardized.")
```

```python
# Verify that no '?' placeholders remain

remaining_question_marks = (df == "?").sum().sum()

print("Remaining '?' values:", remaining_question_marks)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
remaining_question_marks = (df == "?").sum().sum()

print("Remaining '?' values:", remaining_question_marks)
```

```python
# Numerical columns for outlier analysis

numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(list(numerical_columns))
```

```python
# Detect outliers using the IQR method

outlier_summary = {}

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary[column] = len(outliers)

print("Number of outliers in each numerical column:")

for column, count in outlier_summary.items():
    print(f"{column}: {count}")
```

```python
# Visualize numerical variables using boxplots

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

for ax, column in zip(axes.flatten(), numerical_columns):
    sns.boxplot(x=df[column], ax=ax)
    ax.set_title(f"Boxplot of {column}")
    ax.set_xlabel(column)

plt.tight_layout()=
plt.show()
```
